```
using System;

public interface ICreditCard
{
    void Pay(decimal amount);
    void Deposit(decimal amount);
    string GetInfo();
}

public class CreditCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }

    public CreditCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
    }

    public virtual void Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            Console.WriteLine($"Оплата {amount} руб. Успешно! Остаток: {Money} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    public virtual void Deposit(decimal amount)
    {
        Money += amount;
        Console.WriteLine($"Пополнение на {amount} руб. Баланс: {Money} руб.");
    }

    public virtual string GetInfo()
    {
        return $"Карта: {Number} Владелец: {Owner} Баланс: {Money} руб.";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;
}

public class GoldCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    public decimal Cashback { get; set; }

    public GoldCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
        Cashback = 0.05m;
    }

    // Явная реализация интерфейса
    void ICreditCard.Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            decimal cashbackAmount = amount * Cashback;
            Money += cashbackAmount;
            Console.WriteLine($"Оплата {amount} руб. Кэшбэк {cashbackAmount} руб. Баланс: {Money} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    void ICreditCard.Deposit(decimal amount)
    {
        Money += amount;
        Console.WriteLine($"Пополнение на {amount} руб. Баланс: {Money} руб.");
    }

    string ICreditCard.GetInfo()
    {
        return $"Gold карта: {Number} Владелец: {Owner} Баланс: {Money} руб. Кэшбэк: {Cashback * 100}%";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;

    // Собственный метод
    public void ApplyBonus()
    {
        decimal bonus = Money * 0.01m;
        Money += bonus;
        Console.WriteLine($"Начислен бонус: {bonus} руб.");
    }
}

public class PremiumCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    public bool LoungeAccess { get; set; }

    public PremiumCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
        LoungeAccess = true;
    }

    // Явная реализация интерфейса
    void ICreditCard.Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            Console.WriteLine($"Премиум оплата {amount} руб. Успешно! Баланс: {Money} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    void ICreditCard.Deposit(decimal amount)
    {
        decimal bonus = amount * 0.02m;
        Money += amount + bonus;
        Console.WriteLine($"Пополнение {amount} руб. + бонус {bonus} руб. Баланс: {Money} руб.");
    }

    string ICreditCard.GetInfo()
    {
        return $"Premium карта: {Number} Владелец: {Owner} Баланс: {Money} руб. Лаунж: {(LoungeAccess ? "Да" : "Нет")}";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;

    // Собственный метод
    public void UseLounge()
    {
        if (LoungeAccess)
        {
            Console.WriteLine("Добро пожаловать в премиум лаунж!");
        }
    }
}

public class CorporateCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    public string Company { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    public decimal Limit { get; set; }

    public CorporateCard(string number, string owner, string company, string expiryDate, string cvv, decimal limit, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        Company = company;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Limit = limit;
        Money = money;
        IsBlocked = false;
    }

    // Явная реализация интерфейса
    void ICreditCard.Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money + Limit >= amount)
        {
            Money -= amount;
            Console.WriteLine($"Корпоративная оплата {amount} руб. Успешно! Баланс: {Money} руб. Лимит: {Limit} руб.");
        }
        else
        {
            Console.WriteLine($"Превышен лимит!");
        }
    }

    void ICreditCard.Deposit(decimal amount)
    {
        Money += amount;
        Console.WriteLine($"Пополнение корпоративной карты на {amount} руб. Баланс: {Money} руб.");
    }

    string ICreditCard.GetInfo()
    {
        return $"Корпоративная карта: {Number} Владелец: {Owner} Компания: {Company} Баланс: {Money} руб. Лимит: {Limit} руб.";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;

    // Собственный метод
    public void IncreaseLimit(decimal amount)
    {
        Limit += amount;
        Console.WriteLine($"Лимит увеличен на {amount} руб. Новый лимит: {Limit} руб.");
    }
}

// Управление зависимостями
public class BankService
{
    private readonly ICreditCard _card;

    public BankService(ICreditCard card)
    {
        _card = card;
    }

    public void ProcessPayment(decimal amount)
    {
        Console.WriteLine("Банк обрабатывает платеж...");
        _card.Pay(amount);
    }

    public void ShowCardInfo()
    {
        Console.WriteLine(_card.GetInfo());
    }
}


        Console.WriteLine("=== СИСТЕМА БАНКОВСКИХ КАРТ ===\n");

        // Создание карт
        ICreditCard basicCard = new CreditCard("1111-2222", "Иван Иванов", "12/25", "123", 5000);
        ICreditCard goldCard = new GoldCard("3333-4444", "Петр Петров", "06/26", "456", 10000);
        ICreditCard premiumCard = new PremiumCard("5555-6666", "Мария Сидорова", "09/27", "789", 20000);
        ICreditCard corporateCard = new CorporateCard("7777-8888", "Алексей Козлов", "ООО Ромашка", "01/28", "000", 50000, 10000);

        // Использование управления зависимостями
        BankService bank1 = new BankService(basicCard);
        BankService bank2 = new BankService(goldCard);
        BankService bank3 = new BankService(premiumCard);
        BankService bank4 = new BankService(corporateCard);

        Console.WriteLine("ИНФОРМАЦИЯ О КАРТАХ:");
        bank1.ShowCardInfo();
        bank2.ShowCardInfo();
        bank3.ShowCardInfo();
        bank4.ShowCardInfo();

        Console.WriteLine("\nОПЕРАЦИИ:");
        bank1.ProcessPayment(2000);
        bank2.ProcessPayment(3000);
        bank3.ProcessPayment(5000);
        bank4.ProcessPayment(15000);

        Console.WriteLine("\nСОБСТВЕННЫЕ МЕТОДЫ:");
        ((GoldCard)goldCard).ApplyBonus();
        ((PremiumCard)premiumCard).UseLounge();
        ((CorporateCard)corporateCard).IncreaseLimit(10000);
```
