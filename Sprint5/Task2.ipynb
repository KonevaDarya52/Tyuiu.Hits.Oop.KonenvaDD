<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// Делегат для отображения информации
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    // Делегат для обработки событий транспорта
    public delegate void TransportHandler(string message);
    
    // Событие для уведомлений об обновлениях
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}


        // ДЕЛЕГАТЫ
        Console.WriteLine("=== ДЕЛЕГАТЫ ===");
        
        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();
        displayBike();

        // ЛЯМБДА-ВЫРАЖЕНИЯ
        Console.WriteLine("\n=== ЛЯМБДА-ВЫРАЖЕНИЯ ===");
        
        List<Transport> transports = new List<Transport>
        {
            new Car("Model S", "Tesla", 4),
            new Bike("Speedster", "Giant", true),
            new Car("Model 3", "Tesla", 4),
            new Bike("City Bike", "Stels", false)
        };

        // Используем лямбда-выражение для фильтрации транспортных средств
        var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

        Console.WriteLine("Tesla cars:");
        foreach (var teslaCar in teslaCars)
        {
            teslaCar.DisplayInfo();
        }

        // СОБЫТИЯ
        Console.WriteLine("\n=== СОБЫТИЯ ===");
        
        Transport transport = new Transport("Old Model", "Generic");
        
        // Подписываем метод на событие
        transport.TransportUpdated += OnTransportUpdated;
        
        // Обновляем модель; это вызовет событие
        transport.UpdateModel("New Model");
    
    static void OnTransportUpdated(string message)
    {
        Console.WriteLine($"Event received: {message}");
    }
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
using System;
using System.Collections.Generic;
using System.Linq;

// Делегат для отображения информации о животных
public delegate void AnimalDisplayDelegate();

// Делегат для обработки звуков животных
public delegate void AnimalSoundDelegate(string sound);

public class Animal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    // Событие для уведомлений о действиях животных
    public delegate void AnimalActionHandler(string action);
    public event AnimalActionHandler AnimalActionPerformed;

    public Animal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Name: {Name}, Species: {Species}, Age: {Age}");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} makes a sound");
    }

    // Метод для вызова события
    protected void OnAnimalAction(string action)
    {
        AnimalActionPerformed?.Invoke($"{Name} {action}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, string breed, int age)
        : base(name, "Dog", age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Breed: {Breed}");
    }

    public override void MakeSound()
    {
        base.MakeSound();
        OnAnimalAction("barks: Woof-woof!");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} fetches the ball");
        OnAnimalAction("fetched the ball");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, bool isIndoor, int age)
        : base(name, "Cat", age)
    {
        IsIndoor = isIndoor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Is Indoor: {IsIndoor}");
    }

    public override void MakeSound()
    {
        base.MakeSound();
        OnAnimalAction("meows: Meow-meow!");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} purrs softly");
        OnAnimalAction("is purring");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }

    public Bird(string name, bool canFly, int age)
        : base(name, "Bird", age)
    {
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Can Fly: {CanFly}");
    }

    public override void MakeSound()
    {
        base.MakeSound();
        OnAnimalAction("chirps: Tweet-tweet!");
    }

    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} flies in the sky");
            OnAnimalAction("is flying");
        }
        else
        {
            Console.WriteLine($"{Name} cannot fly");
        }
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, string waterType, int age)
        : base(name, "Fish", age)
    {
        WaterType = waterType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Water Type: {WaterType}");
    }

    public override void MakeSound()
    {
        base.MakeSound();
        OnAnimalAction("makes bubbles: Blub-blub!");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} swims in {WaterType} water");
        OnAnimalAction("is swimming");
    }
}


        // ДЕЛЕГАТЫ
        Console.WriteLine("=== РАБОТА С ДЕЛЕГАТАМИ ===");
        
        Dog dog = new Dog("Rex", "Shepherd", 3);
        Cat cat = new Cat("Whiskers", true, 2);
        Bird bird = new Bird("Tweety", true, 1);

        // Создаем делегаты для методов отображения
        AnimalDisplayDelegate displayDog = dog.DisplayInfo;
        AnimalDisplayDelegate displayCat = cat.DisplayInfo;
        AnimalDisplayDelegate displayBird = bird.DisplayInfo;

        // Вызываем методы через делегаты
        Console.WriteLine("Displaying animal info via delegates:");
        displayDog();
        displayCat();
        displayBird();

        // Делегат для звуков животных
        AnimalSoundDelegate soundDelegate = (sound) => 
        {
            Console.WriteLine($"Animal sound: {sound}");
        };

        Console.WriteLine("\nAnimal sounds via delegate:");
        soundDelegate("Woof-woof!");
        soundDelegate("Meow-meow!");
        soundDelegate("Tweet-tweet!");

        // ЛЯМБДА-ВЫРАЖЕНИЯ
        Console.WriteLine("\n=== РАБОТА С ЛЯМБДА-ВЫРАЖЕНИЯМИ ===");
        
        List<Animal> animals = new List<Animal>
        {
            new Dog("Rex", "Shepherd", 3),
            new Dog("Buddy", "Labrador", 2),
            new Cat("Whiskers", true, 2),
            new Cat("Felix", false, 4),
            new Bird("Tweety", true, 1),
            new Bird("Polly", false, 2),
            new Fish("Nemo", "Saltwater", 1)
        };

        // Фильтрация с помощью лямбда-выражений
        var dogs = animals.Where(a => a.Species == "Dog");
        var youngAnimals = animals.Where(a => a.Age < 3);
        var flyingBirds = animals.OfType<Bird>().Where(b => b.CanFly);

        Console.WriteLine("All dogs:");
        foreach (var d in dogs)
        {
            d.DisplayInfo();
        }

        Console.WriteLine("\nYoung animals (age < 3):");
        foreach (var young in youngAnimals)
        {
            young.DisplayInfo();
        }

        Console.WriteLine("\nBirds that can fly:");
        foreach (var flyingBird in flyingBirds)
        {
            flyingBird.DisplayInfo();
        }

        // СОБЫТИЯ
        Console.WriteLine("\n=== РАБОТА С СОБЫТИЯМИ ===");
        
        Dog smartDog = new Dog("Einstein", "Border Collie", 4);
        Cat curiousCat = new Cat("Newton", true, 3);

        // Подписываемся на события
        smartDog.AnimalActionPerformed += OnAnimalAction;
        curiousCat.AnimalActionPerformed += OnAnimalAction;

        // Выполняем действия, которые вызывают события
        Console.WriteLine("Performing animal actions:");
        smartDog.MakeSound();
        smartDog.Fetch();
        
        curiousCat.MakeSound();
        curiousCat.Purr();

        // Лямбда как обработчик события
        Bird singingBird = new Bird("Mozart", true, 1);
        singingBird.AnimalActionPerformed += (action) => 
        {
            Console.WriteLine($"[LAMBDA HANDLER] {action} - and it's beautiful!");
        };

        singingBird.MakeSound();
        singingBird.Fly();
    

    // Обработчик события
    static void OnAnimalAction(string action)
    {
        Console.WriteLine($"[EVENT] {action}");
    }
````
