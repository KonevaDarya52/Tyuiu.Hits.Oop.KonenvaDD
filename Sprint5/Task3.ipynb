```
using System;
using System.Collections.Generic;

public interface ICreditCard
{
    void Pay(decimal amount);
    void Deposit(decimal amount);
    string GetInfo();
}

public class CreditCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    
    // ДОБАВЛЕННЫЕ АТРИБУТЫ
    public string Currency { get; set; }
    public DateTime IssueDate { get; set; }
    public string BankName { get; set; }
    public int PinCode { get; set; }

    // ДЕЛЕГАТ для событий
    public delegate void CardOperationHandler(string message);
    
    // СОБЫТИЯ
    public event CardOperationHandler CardBlocked;
    public event CardOperationHandler PaymentProcessed;

    public CreditCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
        
        // Инициализация новых атрибутов
        Currency = "RUB";
        IssueDate = DateTime.Now;
        BankName = "Default Bank";
        PinCode = 0000;
    }

    public virtual void Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            Console.WriteLine($"Оплата {amount} руб. Успешно! Остаток: {Money} руб.");
            // Вызов события
            PaymentProcessed?.Invoke($"Оплата {amount} {Currency} проведена успешно");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    public virtual void Deposit(decimal amount)
    {
        Money += amount;
        Console.WriteLine($"Пополнение на {amount} руб. Баланс: {Money} руб.");
    }

    public virtual string GetInfo()
    {
        return $"Карта: {Number} Владелец: {Owner} Баланс: {Money} {Currency} Банк: {BankName}";
    }

    public void BlockCard()
    {
        IsBlocked = true;
        CardBlocked?.Invoke($"Карта {Number} заблокирована");
    }
    
    public void UnblockCard() => IsBlocked = false;

    // ДОБАВЛЕННЫЕ МЕТОДЫ
    public void ChangePin(int newPin)
    {
        PinCode = newPin;
        Console.WriteLine($"PIN код изменен для карты {Number}");
    }

    public void DisplayCardAge()
    {
        TimeSpan age = DateTime.Now - IssueDate;
        Console.WriteLine($"Карта выпущена {IssueDate.ToShortDateString()} ({age.Days} дней назад)");
    }

    public virtual void CalculateAnnualFee()
    {
        decimal fee = 500m;
        Console.WriteLine($"Годовая плата: {fee} {Currency}");
    }
}

public class GoldCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    public decimal Cashback { get; set; }
    
    // ДОБАВЛЕННЫЕ АТРИБУТЫ
    public int RewardPoints { get; set; }
    public string CardColor { get; set; }
    public bool Insurance { get; set; }
    public decimal CreditLimit { get; set; }

    // ДЕЛЕГАТ и СОБЫТИЯ
    public delegate void GoldCardHandler(string message);
    public event GoldCardHandler RewardAdded;

    public GoldCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
        Cashback = 0.05m;
        
        // Инициализация новых атрибутов
        RewardPoints = 1000;
        CardColor = "Gold";
        Insurance = true;
        CreditLimit = 50000m;
    }

    // Явная реализация интерфейса
    void ICreditCard.Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            decimal cashbackAmount = amount * Cashback;
            Money += cashbackAmount;
            RewardPoints += (int)(amount * 0.1m);
            Console.WriteLine($"Оплата {amount} руб. Кэшбэк {cashbackAmount} руб. Баланс: {Money} руб.");
            RewardAdded?.Invoke($"Начислено {(int)(amount * 0.1m)} бонусных баллов");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    void ICreditCard.Deposit(decimal amount)
    {
        Money += amount;
        Console.WriteLine($"Пополнение на {amount} руб. Баланс: {Money} руб.");
    }

    string ICreditCard.GetInfo()
    {
        return $"Gold карта: {Number} Владелец: {Owner} Баланс: {Money} руб. Кэшбэк: {Cashback * 100}% Баллы: {RewardPoints}";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;

    // ДОБАВЛЕННЫЕ МЕТОДЫ
    public void ApplyBonus()
    {
        decimal bonus = Money * 0.01m;
        Money += bonus;
        Console.WriteLine($"Начислен бонус: {bonus} руб.");
    }

    public void UseRewardPoints(int points)
    {
        if (points <= RewardPoints)
        {
            RewardPoints -= points;
            decimal cash = points * 0.1m;
            Money += cash;
            Console.WriteLine($"Списано {points} баллов, зачислено {cash} руб.");
        }
    }

    public void DisplayInsuranceInfo()
    {
        Console.WriteLine($"Страховка: {(Insurance ? "Активна" : "Неактивна")}");
    }
}

public class PremiumCard : ICreditCard
{
    public string Number { get; set; }
    public string Owner { get; set; }
    public string ExpiryDate { get; set; }
    public string CVV { get; set; }
    protected decimal Money { get; set; }
    public bool IsBlocked { get; private set; }
    public bool LoungeAccess { get; set; }
    
    // ДОБАВЛЕННЫЕ АТРИБУТЫ
    public bool ConciergeService { get; set; }
    public int PriorityPass { get; set; }
    public decimal InvestmentLimit { get; set; }
    public string MetalType { get; set; }

    // ДЕЛЕГАТ и СОБЫТИЯ
    public delegate void PremiumServiceHandler(string service);
    public event PremiumServiceHandler PremiumServiceUsed;

    public PremiumCard(string number, string owner, string expiryDate, string cvv, decimal money = 0)
    {
        Number = number;
        Owner = owner;
        ExpiryDate = expiryDate;
        CVV = cvv;
        Money = money;
        IsBlocked = false;
        LoungeAccess = true;
        
        // Инициализация новых атрибутов
        ConciergeService = true;
        PriorityPass = 10;
        InvestmentLimit = 100000m;
        MetalType = "Платина";
    }

    // Явная реализация интерфейса
    void ICreditCard.Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("Карта заблокирована!");
            return;
        }

        if (Money >= amount)
        {
            Money -= amount;
            Console.WriteLine($"Премиум оплата {amount} руб. Успешно! Баланс: {Money} руб.");
            PremiumServiceUsed?.Invoke($"Премиум платеж на {amount} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств!");
        }
    }

    void ICreditCard.Deposit(decimal amount)
    {
        decimal bonus = amount * 0.02m;
        Money += amount + bonus;
        Console.WriteLine($"Пополнение {amount} руб. + бонус {bonus} руб. Баланс: {Money} руб.");
    }

    string ICreditCard.GetInfo()
    {
        return $"Premium карта: {Number} Владелец: {Owner} Баланс: {Money} руб. Металл: {MetalType}";
    }

    public void BlockCard() => IsBlocked = true;
    public void UnblockCard() => IsBlocked = false;

    // ДОБАВЛЕННЫЕ МЕТОДЫ
    public void UseLounge()
    {
        if (LoungeAccess && PriorityPass > 0)
        {
            PriorityPass--;
            Console.WriteLine("Добро пожаловать в премиум лаунж!");
            Console.WriteLine($"Осталось визитов: {PriorityPass}");
        }
    }

    public void CallConcierge()
    {
        if (ConciergeService)
        {
            Console.WriteLine("Консьерж-служба подключена. Чем можем помочь?");
            PremiumServiceUsed?.Invoke("Использована услуга консьержа");
        }
    }

    public void DisplayInvestmentOptions()
    {
        Console.WriteLine($"Лимит на инвестиции: {InvestmentLimit} руб.");
    }
}

// Управление зависимостями
public class BankService
{
    private readonly ICreditCard _card;

    public BankService(ICreditCard card)
    {
        _card = card;
    }

    public void ProcessPayment(decimal amount)
    {
        Console.WriteLine("Банк обрабатывает платеж...");
        _card.Pay(amount);
    }

    public void ShowCardInfo()
    {
        Console.WriteLine(_card.GetInfo());
    }
}


    
    {
        Console.WriteLine("=== СИСТЕМА БАНКОВСКИХ КАРТ ===\n");

        // КОЛЛЕКЦИИ
        List<ICreditCard> allCards = new List<ICreditCard>();
        Dictionary<string, ICreditCard> cardDictionary = new Dictionary<string, ICreditCard>();

        // Создание карт
        ICreditCard basicCard = new CreditCard("1111-2222", "Иван Иванов", "12/25", "123", 5000);
        ICreditCard goldCard = new GoldCard("3333-4444", "Петр Петров", "06/26", "456", 10000);
        ICreditCard premiumCard = new PremiumCard("5555-6666", "Мария Сидорова", "09/27", "789", 20000);

        // Добавление в коллекции
        allCards.Add(basicCard);
        allCards.Add(goldCard);
        allCards.Add(premiumCard);
        
        cardDictionary.Add("basic", basicCard);
        cardDictionary.Add("gold", goldCard);
        cardDictionary.Add("premium", premiumCard);

        // ПОДПИСКА НА СОБЫТИЯ
        ((CreditCard)basicCard).PaymentProcessed += (message) => 
            Console.WriteLine($"Событие: {message}");
        ((CreditCard)basicCard).CardBlocked += (message) => 
            Console.WriteLine($"ВНИМАНИЕ: {message}");
            
        ((GoldCard)goldCard).RewardAdded += (message) => 
            Console.WriteLine($"Gold карта: {message}");
            
        ((PremiumCard)premiumCard).PremiumServiceUsed += (service) => 
            Console.WriteLine($"Премиум услуга: {service}");

        // Использование управления зависимостями
        BankService bank1 = new BankService(basicCard);
        BankService bank2 = new BankService(goldCard);
        BankService bank3 = new BankService(premiumCard);

        Console.WriteLine("ИНФОРМАЦИЯ О КАРТАХ:");
        bank1.ShowCardInfo();
        bank2.ShowCardInfo();
        bank3.ShowCardInfo();

        Console.WriteLine("\nОПЕРАЦИИ:");
        bank1.ProcessPayment(2000);
        bank2.ProcessPayment(3000);
        bank3.ProcessPayment(5000);

        Console.WriteLine("\nСОБСТВЕННЫЕ МЕТОДЫ:");
        ((GoldCard)goldCard).ApplyBonus();
        ((GoldCard)goldCard).UseRewardPoints(500);
        ((GoldCard)goldCard).DisplayInsuranceInfo();
        
        ((PremiumCard)premiumCard).UseLounge();
        ((PremiumCard)premiumCard).CallConcierge();
        ((PremiumCard)premiumCard).DisplayInvestmentOptions();
        
        ((CreditCard)basicCard).ChangePin(1234);
        ((CreditCard)basicCard).DisplayCardAge();
        ((CreditCard)basicCard).CalculateAnnualFee();

        Console.WriteLine("\nРАБОТА С КОЛЛЕКЦИЯМИ:");
        Console.WriteLine($"Всего карт в списке: {allCards.Count}");
        
        // Поиск в словаре
        if (cardDictionary.TryGetValue("gold", out ICreditCard foundCard))
        {
            Console.WriteLine($"Найдена карта: {foundCard.GetInfo()}");
        }

        // Перебор коллекции
        Console.WriteLine("\nВсе карты в коллекции:");
        foreach (var card in allCards)
        {
            Console.WriteLine($" - {card.GetInfo()}");
        }

        // Тестирование событий - ИСПРАВЛЕННАЯ СТРОКА
        Console.WriteLine("\nТЕСТИРОВАНИЕ СОБЫТИЙ:");
        ((CreditCard)basicCard).BlockCard();
 
        ICreditCard goldCardInterface = goldCard;
        goldCardInterface.Pay(1000); // Вызовет событие RewardAdded
    }
```
